```python
import pdfplumber as plumber
import pandas as pd
import os
import sys
df_list=list()
empty_pages=list()
header=None
path="/home/shaw/Documents/market_analyzer/assets/sample_data/sample_raw_ads_dataset.pdf"
if not os.path.exists(path):
    print(f"The pdf path : \"{path}\" Does not exists")
else:
    try:
     with plumber.open(path) as f:
         if not f.pages:
             print("PDF is empty")
         else:
             for page_no,page in enumerate(f.pages, start=1):
                 table=page.extract_table()
                 if not table or len(table)<1:
                     empty_pages.append(page_no)
                     continue
                 if header is None:
                     header=table[0]
                     data=table[1:]
                     df_list.append(pd.DataFrame(data,columns=header))
                 else:
                     if table[0]==header: 
                      data=table[1:]
                     else:
                         data=table[:]
                     if data:
                       df_list.append(pd.DataFrame(data,columns=header))
                     else:
                         print("Page only had header")
    except Exception as e:
        exc_type, exc_obj, exc_tb = sys.exc_info()
        print(f"Error in line{exc_tb.tb_lineno}")
        print(f"Critical error : {e}")
c = pd.concat(df_list, ignore_index=True) if df_list else pd.DataFrame()
                     
                     
                     
      
        
```
